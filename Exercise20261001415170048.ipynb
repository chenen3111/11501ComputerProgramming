1. 0.1的二進位表示法為何? 使用markdown語法，列出計算過程和理由，並描述看到的現象。
0.1 的二進位表示法可利用「小數部分連續乘以 2，取整數部分」的方法求得：0.1×2=0.2→0，0.2×2=0.4→0，0.4×2=0.8→0，0.8×2=1.6→1，0.6×2=1.2→1，接著又回到 0.2，因此後面的計算會不斷重複。其結果為：

Plain Text
0.1₁₀ = 0.00011001100110011001100110011...₂

也可寫成：

Plain Text
0.00011(0011)₂
2. 如果是無限多位，Python如何解決？提示：IEEE 754、四捨六入五成雙
如果 0.1 的二進位表示是無限循環小數，電腦不可能用無限多個位元來儲存它，因此 Python 的 float 採用 IEEE 754 Double Precision（雙精度浮點數） 格式來儲存數值。IEEE 754 使用 64 個位元，其中包含 1 個符號位元、11 個指數位元以及 52 個尾數位元，因此只能保留有限長度的二進位數字。當像 0.1 這種無法精確表示的數值被儲存時，系統會截取可儲存的位元數，再依照 IEEE 754 的捨入規則進行近似，而這個規則稱為「四捨六入五成雙（Round Half to Even）」。所謂五成雙，是指當數值剛好落在兩個可表示數值的正中央時，會選擇最後一位為偶數的結果，以減少大量計算時累積的偏差。因此 Python 實際儲存的並不是精確的 0.1，而是最接近 0.1 的一個二進位近似值。這也是為什麼在 Python 中進行浮點數運算時，偶爾會出現像 0.30000000000000004 這樣看似奇怪的結果，其根本原因就在於 IEEE 754 對無限循環二進位小數的近似儲存與捨入機制。

解決方法有幾種：

``

輸出：

或使用格式化輸出：

Python

如果要比較兩個浮點數是否相等，則不建議直接使用 ==，而應使用：

Python
import math
 
print(math.isclose(0.1 + 0.2, 0.3))

輸出：

Plain Text
True

因為 math.isclose() 允許極小的誤差存在，更適合科學計算。若是在金融或會計等需要精確小數運算的情況，則應使用 Decimal：

Python
from decimal import Decimal
 
print(Decimal("0.1") + Decimal("0.2"))

輸出：

Plain Text
0.3
5. 你的方法，對於一般兩個浮點數加法，在科學或金融上是好方法嗎？請思考。
我認為前面使用 round()、格式化輸出或 math.isclose() 的方法，各有適合的情境，但不一定都適合科學或金融上的應用。若只是將結果顯示給使用者看，例如把 0.30000000000000004 顯示成 0.3，使用 round() 或格式化輸出是很方便的方法；然而它們只是把誤差隱藏起來，並沒有真正消除誤差。如果進行大量運算，誤差仍可能持續累積。

在科學計算中，由於測量數據本身就存在誤差，因此通常仍會使用 float，並搭配 math.isclose() 或誤差容忍範圍來比較數值，而不是要求兩個浮點數完全相等。這種做法兼顧了運算速度與可接受的精度，因此是科學與工程領域常見且合理的方法。

在金融、會計或銀行系統中，金額必須精確到分甚至更小單位，任何微小誤差都可能造成問題，因此不適合依賴 float、round() 或 math.isclose()。此時應使用 Decimal 型別，以十進位方式儲存與計算數值，避免二進位浮點數的近似誤差。例如金額計算若使用 Decimal("0.1") + Decimal("0.2")，結果會精確得到 0.3。

因此，對於一般浮點數加法而言，科學計算適合使用 float 搭配 math.isclose()；金融計算則應使用 Decimal 進行精確運算。單純使用 round() 雖然能改善顯示結果，但通常不是解決浮點數精度問題的最佳方法。
