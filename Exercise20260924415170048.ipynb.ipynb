4. Python 為什麼沒有內建的有理數型別？大部分程式設計需求使用整數與浮點數就足夠了，而有理數使用頻率較低，因此 Python 將其放在標準函式庫 fractions 模組中，以 Fraction 類別提供支援。

## 第六部分：AI 模型中的數值表示

人工智慧模型需要儲存大量參數（Parameters）與權重（Weights）。

請查閱資料並回答：

1. AI 模型中的權重通常使用什麼型別儲存？FP64（64 位元浮點數）
FP32（32 位元浮點數）
FP16（16 位元浮點數）
BF16（16 位元 Brain Floating Point）
2. 什麼是 FP64？64 位元浮點數格式，由 1 位符號位、11 位指數位和 52 位尾數位組成。精度最高，約可提供 15 至 16 位有效數字，Python 的 float 通常就是以 FP64 實作。
3. 什麼是 FP32？32 位元浮點數格式，由 1 位符號位、8 位指數位和 23 位尾數位組成。它兼顧精度與運算效率，長期以來是深度學習模型訓練的標準格式。
4. 什麼是 FP16？16 位元浮點數格式，由 1 位符號位、5 位指數位和 10 位尾數位組成。它只需 FP32 一半的記憶體，運算速度較快，但可表示的數值範圍較小，較容易發生溢位或下溢。
5. 什麼是 BF16？16 位元浮點數格式，由 1 位符號位、8 位指數位和 7 位尾數位組成。雖然精度略低，但保留了與 FP32 相近的數值範圍，因此成為現代大型語言模型（LLM）訓練常用的格式。
6. 為什麼大型 AI 模型不直接使用 Python 的 `int` 作為權重？因為 AI 模型的權重通常是小數，例如 0.125、-0.847 或 0.00001，訓練過程中需要不斷進行細微調整，而 int 無法表示小數。此外，Python 的 int 採用任意精度（Arbitrary Precision）設計，記憶體開銷較大，而現代 GPU 對 FP32、FP16 和 BF16 都有專門的硬體加速，因此使用浮點數能同時兼顧精度、記憶體效率與運算速度。

## 第八部分：綜合思考

### 思考題一

Python 官方文件指出：

```python
int
```

具有 Unlimited Precision（無限制精度）。

請回答：

1. Unlimited Precision 是什麼意思？Unlimited Precision（無限制精度） 是指 Python 的 int 不會像許多程式語言的 int32 或 int64 一樣限制只能使用固定的位元數來儲存整數。當整數變大時，Python 可以自動增加所需的儲存空間來保存更多數字位數，因此能表示非常大的整數。
2. 它是否代表可以儲存無限大的整數？不代表真的可以儲存無限大的整數。

Unlimited Precision 的意思是「沒有預先設定固定大小限制」，而不是「無限大」。實際上，Python 整數的大小仍然受到下列因素限制：

電腦記憶體（RAM）
作業系統可用記憶體
執行時間與效能

因此理論上可以非常大，但當記憶體耗盡時仍然會失敗。
3. CPython 如何實現這項特性？在 CPython 中，int 是由 PyLongObject 實作的。與固定大小整數不同，PyLongObject 不只儲存一個機器整數，而是將大整數拆成多個數字區塊（digits）來儲存。當數值變大時，CPython 會配置更多區塊來保存額外的位數。

回答：

1. 為什麼 `int` 幾乎沒有固定上限？因為 Python 的 int 採用 Unlimited Precision（任意精度） 設計，在 CPython 中由 PyLongObject 實作。當整數變大時，Python 不會受限於固定的 32 位元或 64 位元空間，而是會動態配置更多記憶體來儲存額外的數字位數，因此可以表示非常大的整數。真正的限制來自電腦的可用記憶體，而不是型別本身。

### 思考題三

比較下列兩種設計理念：

#### Python 的整數

```text
Arbitrary Precision
正確性優先
```

#### AI 模型常用的 FP16、BF16

```text
Limited Precision
效率優先
```

請說明：

1. 兩種設計理念的差異。Python 的整數（int）採用 Arbitrary Precision（任意精度） 設計，重點是正確性優先。只要記憶體足夠，整數就能持續擴充位數，不會因為數值太大而溢位，因此適合需要精確計算的應用。

AI 模型常用的 FP16、BF16 則採用 Limited Precision（有限精度） 設計，重點是效率優先。它們故意使用較少的位元數來表示數值，以降低記憶體需求並提升 GPU 運算速度，即使因此會損失部分精度。
2. 各自的優點與缺點。優點：

整數計算完全精確。
幾乎不會發生整數溢位。
適合金融、密碼學、科學計算等需要高正確性的領域。

缺點：

數值愈大，記憶體消耗愈多。
大整數運算速度較慢。
不適合大量矩陣運算。
3. 為什麼大型語言模型通常選擇較低精度的表示方式？因為大型語言模型通常擁有數十億甚至數千億個參數。如果每個參數都使用較高精度格式，記憶體與運算成本將大幅增加。使用 FP16 或 BF16 可以將參數儲存空間減半，同時提升 GPU 吞吐量，使模型訓練和推論速度更快。現代研究也發現，神經網路通常能容忍少量數值誤差，因此不一定需要 FP64 等高精度格式。BF16 更因為保留了與 FP32 相近的數值範圍，成為目前大型語言模型訓練的主流選擇。
4. 如果所有 AI 模型都改用 FP64，可能會產生哪些問題？如果全部改用 FP64：

記憶體需求增加為 FP32 的 2 倍、FP16/BF16 的 4 倍。
GPU 顯示卡可容納的模型規模變小。
訓練速度明顯下降。
推論成本提高。
電力消耗增加。
訓練超大型模型所需硬體成本大幅上升。

雖然 FP64 可以提供最高精度，但對大多數 AI 訓練而言，增加的精度通常不足以抵銷其帶來的記憶體與效能成本，因此實務上較少用於大型語言模型的訓練與部署。
